```python
from subprocess import Popen
from glob import glob
import numpy as np


def quant(a, s, m=3):
    a = s / a.sum() * a
    m = np.full(len(a), m, 'B')
    for s in range(s):
        s = a.argmax()
        m[s] -= 1
        if m[s]:
            a[s] -= 1
        else:
            a[s] = 0
        yield s


I = np.array([tuple(i.split('|')) for i in Popen(
    'sqlite3;cards.cdb;select datas.id,alias,type,name from datas join texts on datas.id=texts.id'.split(';'),
    stdout=-1, text=True).stdout.read().split('\n')[:-1]], 'I,I,I,U10')
A = I['f0'].searchsorted(np.where(I['f1'], I['f1'], I['f0']))
E = (I['f2'] & 0x4802040).astype('?')
N = I['f3']
I = I['f0']
W = np.stack([np.add.at(j, A[I.searchsorted(
    (_ := True) and [int(i) for i in open(i).read().split('\n') if (_ := _ and i[:1] != '!') and i.isdigit()])], 1) or j
              for i in
              glob('deck/*.ydk') for j in [np.zeros(len(I), 'f')]], 1)
W = W @ np.square(W).T
W /= W.sum(0) + 1e-45
```

```python
n = 40
i = I.searchsorted([63166095])
a = np.array([4], 'f') / (n + 15)
for p, r in [np.log(n/2)/np.log(4), 0],:
    r = W.std(0) ** r
    s = np.random.rand(len(I)).astype('f')
    _ = 0
    while np.linalg.norm(_ - s) > .1 / (n + 15):
        _ = s
        s = W @ s ** p * r
        s[i] = 0
        s *= (1 - a.sum()) / s.sum()
        s[i] = a
    open('deck/_.ydk', 'wb').write(_:=b'\n'.join(I[list(quant(s * ~E, n)) + list(quant(s * E, 15))].astype('S')))
    # open(f'deck/{"".join(N[i]).replace(" ","")}.ydk', 'wb').write(_)
```
