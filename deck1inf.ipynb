```python
import numpy as np
import torch
from torch.nn import *
from subprocess import Popen
class F(Module):
    def __init__(s, f, *_):
        super().__init__()
        s.f = f
        s._ = _
    def forward(s, x):
        return s.f(x, *s._)
I,m,_ = torch.load('deck1.pk',weights_only=0)
type = np.array([i.split(b'|')for i in Popen('sqlite3;cards.cdb;select id,type from datas'.split(';'),stdout=-1).stdout.read().split()],'I')
type = (type[type[:,0].searchsorted(I),1]&0x4802040).astype('?')
```

```python
x = torch.from_numpy(I.searchsorted([63166095]*3))
ct = np.full(len(I),3,'B')
np.add.at(ct, x, -1)
main = 40-(~type[x]).sum()
ex = 15-type[x].sum()
with torch.no_grad():
    for _ in range(main+ex):
        _ = m(x[None])[0,-1]
        _[ct==0] = -9e9
        if not ex:
            _[type] = -9e9
        elif not main:
            _[~type] = -9e9
        _ = _.argmax()
        ct[_] -= 1
        if type[_]:
            ex -= 1
        else:
            main -= 1
        x = torch.cat([x,_[None]])
open('deck/_.ydk','wb').write(b'\n'.join(I[x].astype('S')))
```
