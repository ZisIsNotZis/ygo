```python
from sqlite3 import connect
import numpy as np
import torch
from torch.nn import *
class Res(Sequential):
    def forward(s, x):
        return super().forward(x)+x
class At(MultiheadAttention):
    def forward(s, x):
        return super().forward(x, x, x, attn_mask=Transformer.generate_square_subsequent_mask(x.size(-2),dtype=torch.float16,device='cuda'), is_causal=True)[0]
class Swiglu(Linear):
    def __init__(s, i, o=0):
        super().__init__(i, 2*(o or i//6*16))
        s._ = Linear(o or i//6*16, i)
    def forward(s, x):
        x,_ = super().forward(x).chunk(2,-1)
        return s._(x*functional.silu(_))
class REnCoder(Module):
    def __init__(s, o, *i):
        super().__init__()
        _ = 0
        for j,i in enumerate(i):
            if isinstance(i,(int,float)):
                setattr(s, str(j), (Embedding if isinstance(i,int)else EmbeddingBag)(int(i),o,scale_grad_by_freq=True))
            else:
                _ += 1
        s._ = Linear(_, o)
    def forward(s, x):
        o = 0
        _ = []
        for i in range(x.size(-1)):
            if isinstance(m:=getattr(s,str(i),None), Embedding):
                o += m(x[...,i])
            elif isinstance(m, EmbeddingBag):
                o += (1<<torch.arange(m.num_embeddings,device=x.device)&x[...,i,None]).sign().to(m.weight.dtype)@m.weight
            else:
                _ += [i]
        return o + s._(x[...,_].to(s._.weight.dtype))
RETRAIN = False
try:
    MAP_ = np.fromfile('model2.i','i').tolist()
    MAP = dict(map(reversed,enumerate(MAP_)))
except FileNotFoundError as _:
    print(_)
    RETRAIN = True
    MAP = {}
ALIAS = {}
CARD = {
    id:(
        ot-(ot>0x4)*0x4,
        (setcode&0xfff).bit_length(),
        setcode>>12&0xf,
        (setcode>>16&0xfff).bit_length(),
        setcode>>28&0xf,
        type,
        atk,
        0 if type&0x4000000 else Def,
        level&15,
        race.bit_length(),
        attribute.bit_length(),
        category,
        level>>24,
        Def-(Def>0x10)*0x10 if type&0x4000000 else 0,
        len(name),
        len(desc),
        sum(bool(j in desc or j in name)<<i for i,j in enumerate('·不为之击在地坏外手方时有特表魔'))
    )
    for id,ot,alias,setcode,type,atk,Def,level,race,attribute,category,name,desc in connect('cards.cdb').execute('select datas.*,name,desc from datas join texts on datas.id=texts.id')
    if ALIAS.setdefault(id,alias or id)and not alias and(RETRAIN or id in MAP)and MAP.setdefault(id,len(MAP))+1
}
CARD = torch.tensor([CARD[i]for i in MAP],device='cuda')
torch.tensor(tuple(MAP),dtype=torch.int).numpy().tofile('model2.i')
```

```python
from glob import glob
from collections import Counter
DECK = [torch.tensor([(MAP[ALIAS[int(i)]],min(n,3))for i,n in Counter(open(i).read().split('!')[0].split()).items()if i.isdigit()and int(i)in ALIAS],dtype=torch.short,device='cuda')for i in glob('deck/**/**.ydk',recursive=True,include_hidden=True)]
DECK = [i for i in DECK if i[...,-1:].sum()>=40]
try:
    if RETRAIN:
        raise FileNotFoundError
    m,o,s = torch.load('model2.pk','cuda',weights_only=False)
except FileNotFoundError as _:
    print(_)
    m = 64
    m = torch.compile(Sequential(
        REnCoder(m,len(MAP),int,3.,475,4.,475,4.,27.,int,int,int,27,8,32.,int,8.,int,int,16.),
        *(Sequential(
            Res(
                LayerNorm(m),
                At(m, 8, batch_first=True),
            ),
            Res(
                LayerNorm(m),
                Swiglu(m),
            ),
        )for _ in range(3)),
        Linear(m, len(MAP))
    ).cuda())
    o = torch.optim.Adam(m.parameters(), .0008)
    s = torch.amp.GradScaler()    
```

```python
while True:
    L = 0
    for _ in range(10000):
        o.zero_grad()
        a = [DECK[i][torch.randperm(len(DECK[i]),dtype=torch.int,device='cuda')]for i in torch.randint(len(DECK),[16],dtype=torch.short)]
        x = torch.stack([torch.constant_pad_nd(torch.cat([i,CARD[i[:,0].int()]],1),(0,0,0,max(map(len,a))-len(i)))for i in a])
        t = torch.zeros(len(x),len(MAP),dtype=torch.half,device='cuda').scatter_add_(1,x[:,:,0],x[:,:,1].half())
        t /= t.sum(1,True)
        with torch.amp.autocast('cuda', torch.float16):
            y = m(x).log_softmax(2)
            l = -y.mean(1).ravel()@t.ravel()
        s.scale(l).backward()
        s.step(o)
        s.update()
        L += l.data.float()
    torch.save((m,o,s),'model2.pk')
    print(L)
```
